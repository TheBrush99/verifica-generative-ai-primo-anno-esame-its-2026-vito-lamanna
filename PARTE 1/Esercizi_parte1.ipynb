# 1️⃣ Esercizio — Embeddings

## Obiettivo
Usare un modello Word2Vec/GloVe **già pre-addestrato** per esplorare le relazioni semantiche tra parole (in inglese).

## Guida passo-passo

**1. Carica il modello.**
Importa `gensim.downloader` e carica un modello pre-addestrato (suggerimento: `api.load("glove-wiki-gigaword-50")` — è piccolo e veloce da scaricare).

> 💬 *Dal notebook di riferimento:* "Carica un modello preaddestrato di Word2Vec" — il modello non lo alleni tu, lo scarichi già pronto e lo interroghi.

*(La cella sotto con `api.load(...)` è già pronta — il download va solo eseguito, non è la parte da esercitare.)*


````python
import gensim.downloader as api


model = api.load("glove-wiki-gigaword-50")
print("Modello caricato. Dimensione di ciascun vettore:", model.vector_size)
````

**2. Trova le parole più simili.**
Scegli una parola inglese presente nel vocabolario e trova le sue 5 parole più simili.

**💡 Hint:** usa il metodo `most_similar()` del modello.


````python
parola="king"
parole_simili=model.most_similar(parola, topn=5)

print(f"Le 5 parole più simili a '{parola}':")
for p, similarita in parole_simili:
    print(f"- {p}: {similarita:.4f}")
````

**3. Confronta due coppie di parole.**
Scegli due coppie di parole — una che ti aspetti *simile*, una che ti aspetti *diversa* — e calcola la similarità tra i due membri di ciascuna coppia. Confronta i due punteggi e commenta il risultato.

**💡 Hint:** usa il metodo `similarity()` del modello.


````python
coppia_1=("cat", "dog")
coppia_2=("cat", "car")
similarità_1=similarita_1=model.similarity(coppia_1[0], coppia_1[1])
similarità_2=similarita_2=model.similarity(coppia_2[0], coppia_2[1])


print(f"Similarità tra 'cat' e 'dog': {similarita_1:.4f}")
print(f"Similarità tra 'cat' e 'car': {similarita_2:.4f}")

if similarita_1 > similarita_2:
    print("Cat e dog sono più simili di cat e car: appartengono entrambi agli animali.")
else:
    print("Il modello assegna a cat e car una similarità maggiore o uguale.")
````


**4. *(Extra facoltativo)* Operazioni aritmetiche tra parole.**

Prova un'operazione tipo `king - man + woman ≈ queen`.

**💡 Hint:** puoi usare nuovamente `most_similar()`, specificando parole positive e negative.


````python
analogia = model.most_similar(positive=['king', 'woman'], negative=['man'], topn=1)
print("king - man + woman:", analogia)
````

# 2️⃣ Esercizio — Una rete neurale

## Obiettivo
Costruire e addestrare da zero una piccola rete neurale ricorrente che, data una parola, preveda quella successiva.

## Guida passo-passo

**1. Scrivi il corpus.**
Scrivi almeno 4-6 frasi brevi (in italiano), con parole che si ripetono tra le frasi: più si ripetono, più in fretta la rete impara.



````python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical
from nltk.tokenize import word_tokenize
import nltk

nltk.download('punkt')
corpus = [
    "Questo è un esempio di generazione di testo",
    "Le reti neurali sono molto potenti per il processing del linguaggio naturale",
    "Le RNN possono essere utilizzate per la previsione della parola successiva",
    "Questo è un esempio semplice di come usare una RNN per generare parole"
]
````

**2. Tokenizza.**
Unisci le frasi, trasformale in minuscolo e tokenizza il testo (`word_tokenize` di NLTK).

**💡 Hint:** prima unisci le frasi con `" ".join(...)`, poi usa `word_tokenize()`.


````python
testo = ' '.join(corpus)
tokens = word_tokenize(testo.lower())
print("Tokens:", tokens)
````

**3. Costruisci il vocabolario.**
Usa `Tokenizer` di Keras per associare un indice numerico a ogni parola.

**💡 Hint:** ti servono `fit_on_texts()` e `word_index`.


````python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([tokens])

total_words = len(tokenizer.word_index) + 1

print("Vocabolario:", tokenizer.word_index)
print("Numero totale di parole:", total_words)
````

**4. Costruisci le coppie di addestramento.**
Per ogni parola del testo, crea la coppia (parola attuale → parola successiva), poi converti le parole in indici numerici e prepara il target per la classificazione.



**5. Costruisci il modello.**
Crea un `Sequential` con `Embedding`, un livello ricorrente e un `Dense` finale.

**💡 Hint:** qui devi ricordare la struttura del modello vista nel notebook LSTM. Parti da questa traccia:

**6. Compila e allena.**
Compila il modello e poi addestralo con i dati preparati.

**💡 Hint:** cerca nel notebook LSTM come vengono usati `compile()` e `fit()`.


**7. Testa il modello.**
Scrivi una funzione che riceva una parola e restituisca la parola prevista dal modello.

**💡 Hint:** la sequenza è: converti la parola → `predict()` → `argmax()` → riconverti l'indice in parola.
```python

    indice = tokenizer.texts_to_sequences([________])[0][0]

```


````python
X = []
y = []

for i in range(len(tokens) - 1):
    parola_prec = tokenizer.texts_to_sequences([tokens[i]])
    parola_succ = tokenizer.texts_to_sequences([tokens[i + 1]])
    if parola_prec and parola_succ and parola_prec[0] and parola_succ[0]:
        X.append(parola_prec[0][0])
        y.append(parola_succ[0][0])

X = np.array(X).reshape(-1, 1)
y = to_categorical(np.array(y), num_classes=total_words)

model = Sequential()
model.add(Embedding(total_words, 50, input_length=1))
model.add(SimpleRNN(32))
model.add(Dense(total_words, activation='softmax'))

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X, y, epochs=100, verbose=1)
````

````python
def predici_parola(parola):
    indice = tokenizer.texts_to_sequences([parola])[0][0]
    pred = model.predict(np.array([[indice]]), verbose=0)
    indice_predetto = np.argmax(pred)
    return tokenizer.index_word[indice_predetto]

print("questo →", predici_parola("questo"))
print("le →", predici_parola("le"))
````

## Obiettivo
Utilizzare un modello Transformer **già pre-addestrato** per svolgere un compito di **summarization**, cioè generare automaticamente un riassunto a partire da un testo più lungo.

> 💬 *Dal notebook di riferimento:* la summarization è un compito **sequence-to-sequence**: il modello riceve una sequenza di testo in ingresso e genera una nuova sequenza di testo in uscita, generalmente più breve.

In questo esercizio utilizzerai **T5**, un Transformer **encoder-decoder**. Il modello è già stato addestrato: non dovrai fare training, ma solo caricare i pesi e utilizzarlo in inferenza.

**3. Carica il modello T5.**
Carica il modello e il tokenizer pre-addestrati.

**💡 Hint:** usa `from_pretrained()` per entrambi.


````python
t5_model = T5ForConditionalGeneration.from_pretrained('t5-small')

t5_tokenizer = T5Tokenizer.from_pretrained('t5-small')
````

**4. Completa la classe `LLMApplications`.**
La classe deve contenere un metodo per la summarization.

**💡 Hint:** nel dizionario `applications` associa il nome dell'applicazione al metodo corrispondente.


````python
class LLMApplications:
    def __init__(self):
        self.applications = {
            'summarization': self.summarization
        }

    def summarization(self, text):
        inputs = t5_tokenizer.encode("summarize: " + text, return_tensors='pt', truncation=True, max_length=512)
        outputs = t5_model.generate(inputs, max_length=150, num_return_sequences=1)
        summary = t5_tokenizer.decode(outputs[0], skip_special_tokens=True)
        return summary
````
